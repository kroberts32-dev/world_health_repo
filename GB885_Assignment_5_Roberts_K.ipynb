##Create new dataframe for countries whose official language is Spanish

```python
# How many rows are in the new dataframe?

# create filter
spanish_language_filter = world_health_df['Official Language'] == 'Spanish'

# apply filter
world_health_df[spanish_language_filter]

#count
world_health_df[spanish_language_filter].shape
```

```python
# mean infant mortality per 1k infants in countires whose official language is Spanish

world_health_df[spanish_language_filter]['Infant mortality_per_1k'].mean()
```

```python
# what is the median maternal mortality in counties whose official langiage is Spanish

world_health_df[spanish_language_filter]['Maternal_Mortality_per_100k'].median()
```

```python
# create GDP per capita field

world_health_df['GDP_per_capita'] = world_health_df['GDP'] / world_health_df['Population']
```
